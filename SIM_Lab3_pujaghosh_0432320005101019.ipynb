```python
#part 1
# 1
import numpy as np
# Generate 3,000 random numbers
random_numbers = np.random.rand(3000)

# Convert values to the range 2 to 12
UniformValues = 2 + (12 - 2) * random_numbers

print(UniformValues)
```

```python
# 2
# Generate 3,000 random values
# Mean = 8 and Standard Deviation = 2

NormalValues = np.random.normal(8, 2, 3000)

print(NormalValues)
```

```python
## 3
# Replace negative values with 0

NormalValues[NormalValues < 0] = 0

print(NormalValues)
```

```python
#Part 2
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Uniform Distribution
plt.subplot(1, 2, 1)

plt.hist(UniformValues, bins=30, alpha=0.5,
         color='blue', edgecolor='black')

plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.title('Uniform Distribution')
plt.grid(True)


# Normal Distribution
plt.subplot(1, 2, 2)

plt.hist(NormalValues, bins=30, alpha=0.5,
         color='blue', edgecolor='black')

plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.title('Normal Distribution')
plt.grid(True)

plt.show()
```

```python
#part 3
# Statistical Analysis of Uniform Distribution

print("Uniform Distribution")

print("Mean:", np.mean(UniformValues))

print("Median:", np.median(UniformValues))

print("Standard Deviation:", np.std(UniformValues))

print("Maximum Wait Time:", np.max(UniformValues))


# Statistical Analysis of Normal Distribution

print("\nNormal Distribution")

print("Mean:", np.mean(NormalValues))

print("Median:", np.median(NormalValues))

print("Standard Deviation:", np.std(NormalValues))

print("Maximum Wait Time:", np.max(NormalValues))
```
