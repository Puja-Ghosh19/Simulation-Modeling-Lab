```python
#task-2
import numpy as np
A = np.array([[11, 2],
              [18, 3]])

print("Matrix A:")
print(A)
# 1. Size of a matrix
size_A = A.shape

# 2. Determinant of a matrix
det_A = np.linalg.det(A)

# 3. Inverse of a matrix
if np.linalg.det(A) != 0:
    inv_A = np.linalg.inv(A)
else:
    inv_A = "Matrix is singular, cannot compute inverse"

# 4. Rank of a matrix
rank_A = np.linalg.matrix_rank(A)

# 5. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Size of A:", size_A)
print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
print("Rank of A:", rank_A)

print("Eigenvalues of A:", eigenvalues)
print("Eigenvectors of A:\n", eigenvectors)
```

```python
#task-4
import numpy as np
B = np.array([[12, 2],
              [18, 3]])

print("\n\nMatrix B:")
print(B)
# 1. Size of a matrix
size_B= B.shape  # Returns dimensions (rows, columns)

# 2. Determinant of a matrix
det_B = np.linalg.det(B)  # Returns the determinant of the matrix

# 3. Inverse of a matrix
if np.linalg.det(B) != 0:  # Check if matrix is invertible
    inv_B = np.linalg.inv(B)  # Compute the inverse of A
else:
    inv_B = "Matrix is singular, cannot compute inverse"  # No inverse if determinant is 0

# 4. Rank of a matrix
rank_B = np.linalg.matrix_rank(B)  # Returns the rank of the matrix

# 5. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(B)  # Compute eigenvalues and eigenvector

print("Size of B:", size_B)
print("Determinant of B:", det_B)
print("Inverse of B:\n", inv_B)
print("Rank of B:", rank_B)

print("Eigenvalues of B:", eigenvalues)
print("Eigenvectors of B:\n", eigenvectors)
```

Comparison & Explanation

1. How did the determinant change and why?

=The determinant changed from -3 to 0 because increasing one element changed the matrix values enough to make the rows linearly dependent.

2. Did the rank change? Explain.

=Yes. The rank changed from 2 to 1 because Matrix B became singular.

3. How did the eigenvalues respond to the value change?

=The eigenvalues changed after modifying one element. This is because eigenvalues depend on all the values of the matrix.

4. Is B easier or harder to invert than A? Why?

=Matrix A is invertible because its determinant is -3 (non-zero). Matrix B is not invertible because its determinant is 0. Therefore, B is harder to invert (actually impossible to invert)
