```python

# PART 1 - SETUP & PERSONALIZATION


import random
import numpy as np
import math


R = 19

maximum = R / 10
radius = R / 10
number_of_letters = 9

# Minimum number of points
minimum_points = 200 * number_of_letters

print("PART 1")
print("R =", R)
print("Range = [0,", maximum, "]")
print("Radius =", radius)
print("Minimum Points =", minimum_points)



# PART 2 - SIMULATION LOGIC

# Check whether a point is inside the circle
def check_point(x, y):

    if x*x + y*y <= radius*radius:
        return True
    else:
        return False


# Generate points and count inside/outside
def simulation(total_points):

    inside = 0
    outside = 0



    for i in range(total_points // 2):

        x = random.uniform(0, maximum)
        y = random.uniform(0, maximum)

        if check_point(x, y):
            inside = inside + 1
        else:
            outside = outside + 1



    for i in range(total_points // 2):

        x = np.random.rand() * maximum
        y = np.random.rand() * maximum

        if check_point(x, y):
            inside = inside + 1
        else:
            outside = outside + 1


    return inside, outside



# PART 3 - OUTPUT & ANALYSIS

# 5 increasing numbers of points
points = [1800, 3600, 5400, 7200, 9000]

print("\nPART 3")

print("Range Used , Total Points , Points Inside , Estimated Pi , Error")



for al_points in points:


    inside, outside = simulation(total_points)

    pi = 4 * inside / total_points
    error = abs(pi - math.pi)


    print(
        "[0,1.9] ",
        total_points,
        "       ",
        inside,
        "        ",
        round(pi, 4),
        "       ",
        round(error, 4)
    )
```
